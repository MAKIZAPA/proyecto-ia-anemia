Exploración estadística de distribuciones, impacto de la altitud andina y correlación de variables.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = '../data/processed/anemia_infantil_peru_endes.csv'
df = pd.read_csv(DATA_PATH)
plt.style.use('seaborn-v0_8-whitegrid')
```

### 1. Distribución de la Variable Objetivo (Anemia)


```python
anemia_counts = df['Anemia'].value_counts()
print(anemia_counts)
plt.figure(figsize=(6, 4))
plt.bar(['No Anémico (70.3%)', 'Anémico (29.7%)'], anemia_counts.values, color=['#2F5496', '#C00000'])
plt.title('Distribución de Clases de Anemia Infantil en el Perú')
plt.ylabel('Cantidad de Niños')
plt.show()
```

### 2. Prevalencia de Anemia según Estratos de Altitud


```python
df['Estrato_Altitud'] = pd.cut(
    df['Altitud_msnm'],
    bins=[-1, 1000, 2500, 3500, 5000],
    labels=['Costa (<1000m)', 'Yunga/Quechua (1000-2500m)', 'Suni (2500-3500m)', 'Puna/Altoandino (>3500m)']
)
tasa_altitud = df.groupby('Estrato_Altitud')['Anemia'].mean() * 100
print(tasa_altitud)
plt.figure(figsize=(8, 4.5))
tasa_altitud.plot(kind='bar', color='#385723')
plt.title('Tasa de Anemia por Piso Altitudinal en el Perú')
plt.ylabel('Prevalencia de Anemia (%)')
plt.xticks(rotation=20)
plt.show()
```

### 3. Matriz de Correlación


```python
features = ['Hemoglobina_Observada', 'Altitud_msnm', 'Edad_Meses', 'Peso_kg', 'Talla_cm', 'Anemia']
plt.figure(figsize=(7, 5))
sns.heatmap(df[features].corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Matriz de Correlación de Pearson')
plt.show()
```
